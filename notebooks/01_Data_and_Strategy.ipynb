# 01 — Data & Strategy

Verify **trainable** InHARD clips (12 industrial actions).

**Never trained / never predicted:** `Assemble system`, `No action`

```python
import sys
from pathlib import Path
NB = Path.cwd() if Path.cwd().name == "notebooks" else Path.cwd() / "notebooks"
sys.path.insert(0, str(NB))

import matplotlib.pyplot as plt
from lib.constants import BLOCKED_ACTIONS, TRAINABLE_ACTIONS, INHARD_DOWNLOAD_URL
from lib.inhard import analyze_training_clips, label_counts, save_step01_summary
from lib.paths import find_inhard_root, list_mock_videos, OUTPUTS_DIR

print("InHARD:", find_inhard_root())
print("Mock videos:", [p.name for p in list_mock_videos()])
```

```python
root = find_inhard_root()
report = analyze_training_clips(root, min_classes=1)
print("Blocked:", list(BLOCKED_ACTIONS))
print("Target actions:", len(TRAINABLE_ACTIONS))
print("On disk — blocked:", report.blocked_on_disk)
print("On disk — trainable:", report.trainable_on_disk)
if report.ok:
    print(f"\n✅ {len(report.clips)} clips, {report.n_classes} classes")
    display(label_counts(report.clips).to_frame("clips"))
else:
    print(f"\n❌ {report.error}")
    print(INHARD_DOWNLOAD_URL)
save_step01_summary(report, root)
```

```python
if report.ok:
    fig, ax = plt.subplots(figsize=(10, 4))
    label_counts(report.clips).plot(kind="bar", ax=ax, color="steelblue")
    ax.set_title("Trainable clips (Assemble + No action excluded)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.savefig(OUTPUTS_DIR / "01_class_balance.png", dpi=120)
    plt.show()
```

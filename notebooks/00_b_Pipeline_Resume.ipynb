# 00_b — Pipeline Resume (cached steps)

Same pipeline as **00**, but **skips embeddings + training** when artifacts already match the current config fingerprint.

| Artifact | Manifest |
|----------|----------|
| `outputs/embeddings.npz` | `outputs/embeddings.manifest.json` |
| `checkpoints/<name>.pt` | `checkpoints/<name>.pipeline.json` |

Change `clips_per_class`, `sample_seed`, `checkpoint_name`, or `train_epochs` → only the affected steps re-run.

Use **00** for a forced full rebuild (`skip_if_cached=False`).

## 1 — Configuration (must match 00)

```python
import sys
from pathlib import Path

NB = Path.cwd() if Path.cwd().name == "notebooks" else Path.cwd() / "notebooks"
sys.path.insert(0, str(NB))

from lib.reload import reload_lib_modules
reload_lib_modules()
from lib.constants import ALL_META_ACTIONS
from lib.pipeline import PipelineConfig
from lib.pipeline_cache import backfill_cache_manifests, cache_status, list_checkpoint_runs
from lib.paths import find_inhard_root

print("InHARD root:", find_inhard_root() or "(not found)")

CFG = PipelineConfig(
    clips_per_class=100,
    max_clips=None,
    exclude_train=(),
    exclude_infer=(),
    min_train_classes=len(ALL_META_ACTIONS),
    sample_seed=42,
    train_epochs=25,
    eval_max_frames=600,
    infer_every=16,
    buffer_frames=32,
    dwell_windows=2,
    skip_if_cached=True,           # ← fingerprint-aware skip
    skip_eval_if_exists=False,     # re-render eval each resume run
    run_data_check=True,
    run_embeddings=True,
    run_train=True,
    run_eval_video=True,
    run_live_app=False,
    checkpoint_name="har_vjepa_all14_100each.pt",
)

# Backfill manifests when npz/ckpt exist but *.manifest.json was never written
backfill = backfill_cache_manifests(CFG)
if backfill.get("embeddings_backfilled") or backfill.get("train_backfilled"):
    print("Backfilled cache manifests:", backfill)

status = cache_status(CFG)
print("\nCache status:")
for k, v in status.items():
    print(f"  {k}: {v}")

if status["embeddings_valid"]:
    print("\n✓ Embeddings — skip step 02")
else:
    print("\n→ Embeddings — will extract (missing or config changed)")
if status["train_valid"]:
    print("✓ Checkpoint — skip step 03")
else:
    print("→ Train — will run (missing or config changed)")

runs = list_checkpoint_runs()
if runs:
    print("\nSaved checkpoint runs:")
    for r in runs:
        print(f"  {r['checkpoint']}  epochs={r.get('epochs')}  {r.get('created_at', '')[:19]}")
```

## 2 — Run pipeline (resume)

```python
reload_lib_modules()
from lib.pipeline import run_pipeline

results = run_pipeline(CFG)
print("status:", results["status"])
for step, data in results.get("steps", {}).items():
    skipped = data.get("skipped") or data.get("reason") == "cache_hit"
    tag = "SKIP" if skipped else "RUN"
    print(f"  [{tag}] {step}: {data}")
results
```

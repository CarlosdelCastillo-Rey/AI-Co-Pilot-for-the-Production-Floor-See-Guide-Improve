# 05 — Live Camera Demo (Per-Person UI)

OpenCV window with:
- **ByteTrack** `#track_id` per person
- **Individual** action label (crop buffer → V-JEPA2 → MLP)
- **Memory:** dwell smoothing + JSONL log per session

```bash
cd notebooks
python lib/live_app.py --video ../data_sample/mock-videos/Industrial-One.mp4
python lib/live_app.py --webcam 0
```

**Keys:** `q` quit · `space` pause · `r` reset tracks

````python
import subprocess
import sys
from pathlib import Path

NB = Path.cwd() if Path.cwd().name == "notebooks" else Path.cwd() / "notebooks"
sys.path.insert(0, str(NB))
from lib.paths import CHECKPOINTS_DIR, list_mock_videos

ckpt = CHECKPOINTS_DIR / "har_vjepa_mlp.pt"
if not ckpt.is_file():
    raise FileNotFoundError("Run 03_Train_HAR_Head.ipynb first")

cmd = [sys.executable, str(NB / "lib" / "live_app.py"), "--checkpoint", str(ckpt)]
vids = list_mock_videos()
if vids:
    cmd.extend(["--video", str(vids[0])])
else:
    cmd.extend(["--webcam", "0"])
print(" ".join(cmd))
subprocess.run(cmd, cwd=str(NB))
````
